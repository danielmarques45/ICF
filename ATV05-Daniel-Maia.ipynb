```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
from scipy.interpolate import CubicSpline
from scipy.integrate import simpson
from numpy import trapezoid
from sympy import symbols, integrate, sin, cos, sqrt

s = symbols('s') # variável global que utilizaremos em todas as integrações

def dominio(inicio,fim):
    return np.linspace(inicio,fim,52)[1:51] # excluindo 'inicio' e 'fim' para evitar problemas de domínio
def dom_itp(x): # ampliação do domínio para as interpolações
    return np.linspace(min(x),max(x),250)
def ruido(y):
    return 0.06*(1-2*(np.random.rand(len(y))))


def gera_grafico(x,y,titulo):
    plt.figure(figsize=(6,6))
    plt.title(titulo)
    plt.xlabel("$x$", fontsize=14)
    plt.ylabel("$y$", fontsize=14)
    plt.plot(x,y,'x', markersize=11)
    cs = CubicSpline(x,y)
    x_itp = dom_itp(x)
    plt.plot(x_itp,cs(x_itp),'--')
    plt.legend(['Dados', 'Cubic spline'])
    plt.show()

# Definindo as funções que usaremos em todos os exercícios e já imprimem o valor que buscamos
def integral_definida(funcao,x):
    area_exata = integrate(funcao,(s,min(x),max(x))) # como as bordas do domínio de cada função serão os limites de integração, podemos utilizar min(x) e max(x) 
    print(f"Valor exato (SymPy): {area_exata.evalf()}")
def trapz(x,y):
    area = trapezoid(y,x)
    print ('Trapézio - área = ',area)
def simps(x,y):
    area = simpson(y,x)
    print ('Simpson - área = ',area)
```

```python
# Atividade 1 - Para a função y = 2.0*sin^2(3.0*x + 1.77)*cos(1.0*x), no intervalo de [0,2PI] faça:
# (a) gere um conjunto de dados incluindo ruído;
# (b) Faça uma suavização ou interpolação (com resultados em gráfico);
# (c) Calcule o valor da integral pelos métodos do trapézio e Simpson;
# (d) Compare com o valor exato da integral;

# Todas as funções vão ser definidas de forma simbólica para facilitar a integração exata, além da atribuição da variável yi

x1 = dominio(0,2*np.pi)
x1a = dom_itp(x1) # definindo x1 ampliado que será utilizado para aumentar a precisão das aproximações
func1 = 2*((sin(3*s + 1.77))**2)*cos(s)
y1 = 2*((np.sin(3*x1 + 1.77))**2)*np.cos(x1)
y1 += ruido(y1)
y1a = 2*((np.sin(3*x1a + 1.77))**2)*np.cos(x1a) # y1 ampliado divide a função em mais pontos, melhorando a precisão de trapz e simps
y1a += ruido(y1a)

gera_grafico(x1,y1,'Exercício 1') # gera o gráfico com os dados e a suavização Cubic Spline
integral_definida(func1,x1)
trapz(x1a,y1a)
simps(x1a,y1a)
```

```python
# Atividade 2 - Para a função y = 1/[x^2 * sqrt(4 - x^2)];, no intervalo de [0,2] faça:
# (a) gere um conjunto de dados incluindo ruído;
# (b) Faça uma suavização ou interpolação (com resultados em gráfico);
# (c) Calcule o valor da integral pelos métodos do trapézio e Simpson;
# (d) Compare com o valor exato da integral;

x2 = dominio(0, 2)
x2a = dom_itp(x2)
func2 = 1/((s**2) * sqrt(4 - s**2))
y2 = 1/((x2**2) * np.sqrt(4 - x2**2))
y2 += ruido(y2)
y2a = 1/((x2a**2) * np.sqrt(4 - x2a**2))
y2a += ruido(y2a)

gera_grafico(x2,y2,'Exercício 2') # gera o gráfico com os dados e a suavização Cubic Spline
integral_definida(func2,x2)
trapz(x2a,y2a)
simps(x2a,y2a)
```

```python
# Atividade 3 - Para a função y = 1/sqrt(x^2 - 64), no intervalo de [8,20] faça:
# (a) gere um conjunto de dados incluindo ruído;
# (b) Faça uma suavização ou interpolação (com resultados em gráfico);
# (c) Calcule o valor da integral pelos métodos do trapézio e Simpson;
# (d) Compare com o valor exato da integral;

x3 = dominio(8, 20)
x3a = dom_itp(x3)
func3 = 1/sqrt(s**2 - 64)
y3 = 1/np.sqrt(x3**2 - 64)
y3 += ruido(y3)
y3a = 1/np.sqrt(x3a**2 - 64)
y3a += ruido(y3a)

gera_grafico(x3,y3,'Exercício 3') # gera o gráfico com os dados e a suavização Cubic Spline
integral_definida(func3,x3)
trapz(x3a,y3a)
simps(x3a,y3a)
```

```python
# Atividade 4 - Para a função y = cos^2(x^2), no intervalo de [0,20] faça:
# (a) gere um conjunto de dados incluindo ruído;
# (b) Faça uma suavização ou interpolação (com resultados em gráfico);
# (c) Calcule o valor da integral pelos métodos do trapézio e Simpson;
# (d) Compare com o valor exato da integral;

x4 = dominio(0, 20)
x4a = dom_itp(x4)
func4 = (cos(s**2))**2
y4 = (np.cos(x4**2))**2
y4 += ruido(y4)
y4a = (np.cos(x4a**2))**2
y4a += ruido(y4a)

gera_grafico(x4,y4,'Exercício 4') # gera o gráfico com os dados e a suavização Cubic Spline
integral_definida(func4,x4)
trapz(x4a,y4a)
simps(x4a,y4a)
```
