##### dicmes(nome)

```python
def dicmes(nome):
    '''
    Função que recebe o nome do arquivo das médias mensais e armazena os dados num dicionário, que é retornado ao fim.
    ---------------------
    Parâmetro: nome (STR)
    ---------------------
    Retorno: dicanomes (DICT)
    '''
    dicanomes = {}
    try:
        with open(nome) as orig:
            for linha in orig:
                dados = linha.split() 
                #percorrendo o arquivo linha por linha e gerando uma lista, mais fácil de trabalhar que strings
                ano = int(dados[0])     
                mes = int(dados[1])  
                média = float(dados[3])
                # criando a chave do ano, caso não exista
                if ano not in dicanomes:
                    dicanomes[ano] = {}  
                # não é um bloco else, pois será executado para todas as linhas, depois de garantir que a chave existe
                dicanomes[ano][mes]= média
    except FileNotFoundError:  
        print("Arquivo não encontrado!")
    return dicanomes

basededadosMENSAIS = dicmes('SN_m_tot_V2.0.txt')
```

##### dicdia(nome)

```python
def dicdia(nome):
    '''
    Função que recebe o nome do arquivo das manchas diárias e armazena os dados num dicionário, que é retornado ao fim.
    ---------------------
    Parâmetro: nome (STR)
    ---------------------
    Retorno: dicanomesdia (DICT)
    '''
    dicanomesdia = {}
    try:
        with open(nome) as orig:
            for linha in orig:
                dados = linha.split() #percorrendo o arquivo linha por linha e gerando uma lista, mais fácil de trabalhar que strings
                ano = int(dados[0])     
                mes = int(dados[1])  
                dia = int(dados[2]) 
                manchas = int(dados[4])
                # criando a chave do ano e do mês, caso não existam
                if ano not in dicanomesdia:
                    dicanomesdia[ano] = {}
                if mes not in dicanomesdia[ano]:
                    dicanomesdia[ano][mes] = {}
                # não é um bloco else, pois será executado para todas as linhas, depois de garantir que as chaves existem
                dicanomesdia[ano][mes][dia] = manchas
    except FileNotFoundError:
        print("Arquivo não encontrado!")
        
    return dicanomesdia
        
basededadosDIARIOS = dicdia('SN_d_tot_V2.0.txt')
```

##### verif_dataexiste(data)

```python
def verif_dataexiste(data):
    '''
    Função que recebe uma data de qualquer tipo em qualquer formato e verifica se está no tipo e no formato correto 
    e se a data é válida. Para facilitar a impressão da mensagem de erro em funções posteriores, a saída é uma lista
    de um elemento BOOL se a data for válida, e se for inválida, uma lista com 3 elementos: o primeiro indica que é 
    inválida, o segundo aponta qual parte está errada e o terceiro, a mensagem para que o usuário corrija a entrada.
    ---------------------
    Parâmetro: data (STR)  
    ---------------------
    Retorno: saida (LIST) lista que engloba:
             veri (BOOL) resultado da verificação.
             erro (STR) erro ocorrido.
             mensagem (STR) instrução para a correção do erro.
    '''
    veri = False
    if type(data) is not str:
        erro = 'Tipo inválido!'
        mensagem = 'Insira a data no formato "dd/mm/aaaa", com as aspas.'
    elif len(data)!=10:
        erro = 'Formato inválido!'
        mensagem = 'Insira a data no formato "dd/mm/aaaa", com as aspas e as barras.'
    else:
        try:
            dia = int(data[:2])
            mes = int(data[3:5])
            ano = int(data[6:10])
        except ValueError:
            erro = 'Formato inválido!'
            mensagem = 'Insira a data no formato "dd/mm/aaaa", usando apenas números e barras.'
            saida = [veri, erro, mensagem]
            return saida
        if dia<1 or dia>31:
            erro ='Dia inválido!'
            mensagem = 'Certifique-se que não inseriu um dia menor que 1 ou maior que 31.'
        elif mes<1 or mes>12:
            erro ='Mês inválido!'
            mensagem = 'Certifique-se que não inseriu um mês menor que 1 ou maior que 12.'
        elif mes in (4,6,9,11) and dia==31:
            erro ='Dia inválido!'
            mensagem = 'Esse mês não tem dia 31.'
        elif mes==2 and dia>29:
            erro ='Dia inválido!'
            mensagem = 'Fevereiro não tem mais que 29 dias.'
        elif (ano%4!=0 and ano%400!=0) and (mes==2 and dia==29): #dia 29 de fevereiro de um ano não bissexto não existe
            erro = 'Dia inválido!'
            mensagem = 'Esse ano não é bissexto, então fevereiro não tem 29 dias.'
        else: #não há problemas
            veri = True
            erro = ''
            mensagem = ''
    saida = [veri, erro, mensagem]
    return saida
```

##### verif_duasdatas(data1,data)

```python
def verif_duasdatas(data1,data2):
    '''
    Dadas duas datas de entrada, verifica se são strings, elas são válidas, se a segunda é posterior à primeira e 
    se elas não são subsequentes. Se esses critérios forem atingidos, retorna uma lista [True,]. Se não, informa esse resultado,
    o erro ocorrido e orienta para sua correção.
    ---------------------
    Parâmetro: data1,data2 (ANY TYPE) datas inicial e final.
    ---------------------
    Retorno: saida (LIST) lista que engloba:
             veri (BOOL) resultado da verificação.
             erro (STR) erro ocorrido.
             mensagem (STR) instrução para a correção do erro.
    '''
    veri = False
    if verif_dataexiste(data1)[0]==False: #checando tipo da entrada e se a data existe
        erro = 'Primeira data inválida!'
        mensagem = verif_dataexiste(data1)[2]
    elif verif_dataexiste(data2)[0]==False:
        erro = 'Segunda data inválida!'
        mensagem = verif_dataexiste(data2)[2]
    else:
        # garantindo que o ano tem dados registrados------------
        ano1 = int(data1[6:10])
        ano2 = int(data2[6:10])
        if verifanoin_DIARIOS(ano1)[0]==False:
            erro = 'Primeira data inválida!'
            mensagem = verifanoin_DIARIOS(ano1)[2]
        elif verifanoin_DIARIOS(ano2)[0]==False:
            erro = 'Segunda data inválida!'
            mensagem = verifanoin_DIARIOS(ano2)[2]
        else:
            #garantido que a segunda data é posterior à primeira e que não são meses subsequentes------------
            dia1 = int(data1[:2])
            dia2 = int(data2[:2])
            mes1 = int(data1[3:5])
            mes2 = int(data2[3:5])
            if (ano1>ano2) or (ano1==ano2 and mes1>mes2) or (ano1==ano2 and mes1==mes2 and dia1>dia2):
                erro = 'Ordem cronológica inválida!'
                mensagem = 'Insira a segunda data sendo posterior à primeira.'
            elif (ano1==ano2 and (mes2==mes1+1 or mes1==mes2)) or (ano2==ano1+1 and mes1==12):
                erro = 'Não há um mês entre essas duas datas para ser analisado!'
                mensagem = 'Insira meses que não são subsequentes.'
            else: # entradas válidas
                veri = True
                erro = ''
                mensagem = ''
    saida = [veri,erro,mensagem]
    return saida
```

##### somadomes(mes,ano)

```python
def somadomes(mes,ano):
    '''
    Dado um mês e um ano de entrada, soma a quantidade de manchas registradas em cada dia daquele mês.
    ---------------------
    Parâmetro: mes,ano (INT)
    ---------------------
    Retorno: soma (INT)
    '''
    dadoslimpos = []
    diassemdados = 0 # será importante para calcular a média
    for elem in basededadosDIARIOS[ano][mes].items(): #deixando de contar os dias em que não há dados registrados
        if elem[1]==-1:
            dadoslimpos += [(elem[0],0)]
            diassemdados += 1
        else: 
            dadoslimpos += [elem]
    soma  = sum([dado[1] for dado in dadoslimpos])
    diascomdados = len(basededadosDIARIOS[ano][mes])-diassemdados
    return soma,diascomdados

```

##### diassemmanchaano(ano)

```python
def diassemmanchaano(ano): 
    '''
    Dado um ano de entrada, conta, em cada mês, em quantos dias não foram observadas manchas.
    Retorna uma lista com os meses e seus valores, e uma mensagem para o usuário.
    ---------------------
    Parâmetro: ano (INT)
    ---------------------
    Retorno: listafinal (LIST) é a lista com todos os meses e suas contagens, no formato [('mm/aaaa',contagem)]
             MSGcontagemtotal (STR) é uma mensagem que informa o usuário do resultado da função.
    '''
    #
    # QUANDO FOR CHAMAR NO MENU, TEM QUE FAZER IMPRIMIR!!!
    #
    listafinal = None
    MSGcontagemtotal = None
    if type(ano) is not int:
        print('Insira um ano inteiro, sem aspas.')
    elif verifanoin_DIARIOS(ano)[0]==False:
       print(f'{" ".join(verifanoin_DIARIOS(ano)[1:])}')
    else:
        MSGcontagemtotal = f'Dias sem manchas em cada mês de {ano}:\n' #indicará os mês e suas contagens
        listafinal = [] #dará o resultado numérico
        for mes in range(1,13): 
            contdias = basededadosDIARIOS[ano][mes].items() #obtendo uma lista de tuplas para contar os dias sem manchas
            contagem = 0
            for dados in contdias: #contando os dias
                if dados[1]==0: 
                    contagem+=1
            listafinal+= [(f'{mes:>02}/{ano}',contagem)]
            meses = ['Janeiro', 'Fevereiro', 'Março', 'Abril', 'Maio', 'Junho', 'Julho', 'Agosto', 'Setembro', 'Outubro', 'Novembro', 'Dezembro']
            MSGcontagemtotal += f'{meses[mes-1]}: {contagem} dias\n'
        MSGcontagemtotal = MSGcontagemtotal.rstrip() #removendo o último \n
    return listafinal,MSGcontagemtotal
```

##### maximoentredatas(data1,data2)

```python
def maximoentredatas(data1,data2):
    '''
    Dadas duas datas distintas, encontra o(s) ano(s) o(s) mês(es) entre elas a maior quantidade de manchas observadas.
    Retorna uma lista com os meses e qual foi a maior contagem, além de uma mensagem para o usuário.
    ---------------------
    Parâmetro: data1,data2 (STR,STR) datas inicial e final. Até que ambas estejam no formato dd/mm/aaaa, a função 
               solicita a reinserção dos dados.
    ---------------------
    Retorno: listamaximos (LIST) é a lista com todos os meses que tiveram a maior quantidade de manchas e qual foi 
             essa quantidade, no formato [('mm/aaaa',quantidade)]
             saidatexto (STR) é uma mensagem que informa o usuário do resultado da função.
    '''
    if verif_duasdatas(data1,data2)[0]==False: # prevenindo quaisquer invalidade nas datas
        print(f'{" ".join(verif_duasdatas(data1,data2)[1:])}')
    else: # entradas válidas
        mes1 = int(data1[3:5])
        mes2 = int(data2[3:5])
        ano1 = int(data1[6:10])
        ano2 = int(data2[6:10])
        # contando a quantidade de manchas------------
        listadcont = []
        if ano1==ano2:
            for mes in range(mes1,mes2):
                contagem = somadomes(mes,ano1)[0]
                listadcont +=  [(f'{mes:>02}/{ano1}',contagem)]
        else: 
            #ano2>ano1. contando primeiro os anos dados na entrada, pois seu tratamento será diferente dos anos intermitentes
            for mes in range(mes1,13):
                contagem = somadomes(mes,ano1)[0]
                listadcont +=  [(f'{mes:>02}/{ano1}',contagem)]
            for mes in range(1,mes2):
                contagem = somadomes(mes,ano2)[0]
                listadcont +=  [(f'{mes:>02}/{ano2}',contagem)]
            for ano in range(ano1+1,ano2):
                for mes in range(1,13):
                    contagem = somadomes(mes,ano)[0]
                    listadcont +=  [(f'{mes:>02}/{ano}',contagem)]
        # encontrado a(s) maior(es) contagem(ns)------------
        maior_parcial = -1 
        indicedosmaiores=[] 
        for i in range(len(listadcont)):
            if listadcont[i][1]<maior_parcial: #quantidade menor que o maior parcial, não nos interessa
                pass
            elif listadcont[i][1]==maior_parcial: #igual ao maior parcial, vamos "anotar" seu índice
                indicedosmaiores += [i]
            elif listadcont[i][1]>maior_parcial:
                maior_parcial = listadcont[i][1]
                indicedosmaiores= [i]
        # criando a saída para o usuário------------
        nomedosmeses = ['Janeiro', 'Fevereiro', 'Março', 'Abril', 'Maio', 'Junho', 'Julho', 'Agosto', 'Setembro', 'Outubro', 'Novembro', 'Dezembro']
        if len(indicedosmaiores)==1:
            ind = indicedosmaiores[0] #acessando na lista qual foi o mês e o ano e obtendo seus valores numéricos
            nummes = int(listadcont[ind][0][:2]) 
            ano = int(listadcont[ind][0][3:])
            mes = nomedosmeses[nummes-1] #obtendo o mês por extenso
            saidatexto = f'Entre essas datas, o mês com mais manchas foi {mes} de {ano}, com {maior_parcial} manchas.'
            listamaximos = listadcont[ind]
        else: #vários meses com a mesma quantia; devemos imprimir todos
            todos_os_meses =''
            listamaximos = []
            for ind in indicedosmaiores:
                nummes = int(listadcont[ind][0][:2]) 
                ano = int(listadcont[ind][0][3:])
                mes = nomedosmeses[nummes-1]
                listamaximos += listadcont[ind]
                todos_os_meses += f'{mes} de {ano}, '
            saidatexto = f'Entre essas datas, os meses com mais manchas foram {todos_os_meses}com {maior_parcial} manchas'
        return listamaximos,saidatexto
```

##### maxmindatas(data1,data2)

```python
def maxmindatas(data1,data2):
    '''
    Dadas duas datas distintas, determina a quantidade máxima e mínima de manchas solares observadas em um mês entre elas.
    ---------------------
    Parâmetro: data1, data2 (STR, STR) datas inicial e final. Até que ambas estejam no formato dd/mm/aaaa, a função 
               solicita a reinserção dos dados.
    ---------------------
    Retorno: minmax (TUPLE) tupla contendo o valor mínimo absoluto e o valor máximo absoluto encontrados, nessa ordem.
             saidatexto (STR) é uma mensagem que informa o usuário do resultado da função.
    ''' 
    if verif_duasdatas(data1,data2)[0]==False: # prevenindo quaisquer invalidade nas datas
        print(f'{" ".join(verif_duasdatas(data1,data2)[1:])}')
    else:
        minimo = minimoentredatas(data1, data2)[0][1]
        maximo = maximoentredatas(data1, data2)[0][1]
        minmax = minimo,maximo
        saidatexto = f'Entre {data1} e {data2}, o menor número de manchas observadas em um mês foi {minimo} e o maior foi {maximo}.'
        return minmax,saidatexto
```

##### mediamensal_ano(ano)

```python
def mediamensal_ano(ano):
    '''
    Dado um ano de entrada, calcula e organiza a média mensal de manchas solares para cada um de seus meses. Perceba que a saída é
    idêntica aos dados contidos no arquivo de médias mensais.
    ---------------------
    Parâmetro: ano (INT) ano a ser analisado.
    ---------------------
    Retorno: listamedias (LIST) lista contendo as médias mensais calculadas no formato [((mes,ano), media)].
             retornar a tupla (mes,ano) ao invés de ('mm/aaaa') nos ajudará a calcular a média suavizada.
             saidatexto (STR) é uma mensagem formatada que mostra as médias de todos os meses do ano.
    '''
    if verifanoin_DIARIOS(ano)[0]==False:
        print(f'{" ".join(verifanoin_DIARIOS(ano)[1:])}')
    else:
        # calculando a média ------------
        mediadosmeses = []
        for mes in range(1,13):
            diascomdados = somadomes(mes,ano)[1]
            if diascomdados > 0:
                media = somadomes(mes,ano)[0]/diascomdados
            else: # não há dados registrados em nenhum dos dias
                media = 0
            mediadosmeses += [((mes,ano),round(media,1))]
        # criando a saída ------------
        meses = ['Janeiro', 'Fevereiro', 'Março', 'Abril', 'Maio', 'Junho', 'Julho', 'Agosto', 'Setembro', 'Outubro', 'Novembro', 'Dezembro']
        saidatexto = f'Média de cada mês de {ano}:\n'
        for (mes,_),media in mediadosmeses:
            saidatexto += f'{meses[mes-1]}: {media}\n'
        saidatexto = saidatexto.rstrip() #removendo o último \n
        return mediadosmeses,saidatexto
```

##### medsvz(ano)

```python
def medsvz(ano):
    '''
    Dado um ano de entrada, calcula a média suavizada de todos os meses daquele ano. Pressupõe que a entrada é válida.
    Essa função não será diretamente chamada pelo usuário, mas escolhi colocar ela logo antes da média suavizada entre duas datas
    para facilitar a compreensão.
    ---------------------
    Parâmetro: ano (INT)
    ---------------------
    Retorno: listamedsvz (LIST) lista no formato[((mes,ano),mediasuavizada)]
    '''
    listamedsvz = []
    # calculando para os meses de janeiro a junho, que precisam das médias do ano anterior
    for mes in range(1,7):
        # mês em que vamos começar a contar no ano anterior
        inicanoanterior = 6 + mes 
        # médias do ano anterior 
        medanoant = [0.5*basededadosMENSAIS[ano-1][inicanoanterior]] + [basededadosMENSAIS[ano-1][mesdoant] for mesdoant in range(inicanoanterior+1,13)]  
        medano = [0.5*basededadosMENSAIS[ano][mes+6]] + [basededadosMENSAIS[ano][mesdoano] for mesdoano in range(1,mes+6)]
        mdsvz_dessemes = round((sum(medano)+sum(medanoant))/12,1)
        listamedsvz += [((mes,ano), mdsvz_dessemes)]
    # calculando para os meses de julho a dezembro, que precisam das médias do ano seguinte. variáveis análogas às do procedimento acima
    for mes in range(7,13):
            fimanoseg = mes - 6
            medano = [0.5*basededadosMENSAIS[ano][mes-6]] + [basededadosMENSAIS[ano][mesdoano] for mesdoano in range(mes-5,13)]
            medanoseg = [0.5*basededadosMENSAIS[ano+1][fimanoseg]] + [basededadosMENSAIS[ano+1][mesdoseg] for mesdoseg in range(1,fimanoseg)]
            medsvz_dessemes = round((sum(medano)+sum(medanoseg))/12,1)
            listamedsvz += [((mes,ano), medsvz_dessemes)]
    return listamedsvz
```

```python
def picos_locais_medsvz(ano1, ano2):
    '''
    Identifica os picos locais da média suavizada de manchas solares no período determinado.
    ---------------------
    Parâmetro: ano1 (INT), ano2 (INT) anos inicial e final do período.
    ---------------------
    Retorno: picos (LIST) lista contendo tuplas com o índice cronológico e o mês/ano de cada pico [(índice, (mes, ano))].
    '''
    # verificando a validade das entradas com a base de dados MENSAIS ------------
    if verif_doisanosMENSAIS(ano1, ano2)[0]==False:
        print(f'{" ".join(verif_doisanosMENSAIS(ano1,ano2)[1:])}')
        return None
    elif ano1==1749:
        print('Não é possível calcular a média suavizada dos 6 primeiros meses de 1749, então é preciso inserir um ano posterior a ele. Tente novamente.')
        return None
    elif ano2==2025:
        print('Não é possível calcular a média suavizada de dezembro de 2025, então é preciso inserir um ano anterior a ele. Tente novamente.')
        return None
    else: 
        # obtendo a lista de médias suavizadas ------------
        listasuavizada = mediasuavizadaperiodo(ano1, ano2)[0]
        valores = [item[1] for item in listasuavizada]
        
        # identificando os índices e datas dos picos locais ------------
        picos = []
        for i in range(1, len(valores) - 1):
            if valores[i] > valores[i-1] and valores[i] > valores[i+1]:
                picos += [(i, listasuavizada[i][0])]
                # se um mês é maior que seu antecessor e que seu sucessor, é um pico local. guardamos o índice 'i' porque a posição na lista funciona como uma linha do tempo em meses, simplificando as contas
        return picos
```

```python
def vales_locais_medsvz(ano1, ano2):
    '''
    Identifica os vales locais da média suavizada de manchas solares no período determinado.
    ---------------------
    Parâmetro: ano1 (INT), ano2 (INT) anos inicial e final do período.
    ---------------------
    Retorno: vales (LIST) lista contendo tuplas com o índice cronológico e o mês/ano de cada vale [(índice, (mes, ano))].
    '''
    # verificando a validade das entradas com a base de dados MENSAIS ------------
    if verif_doisanosMENSAIS(ano1, ano2)[0]==False:
        print(f'{" ".join(verif_doisanosMENSAIS(ano1,ano2)[1:])}')
        return None
    elif ano1==1749:
        print('Não é possível calcular a média suavizada dos 6 primeiros meses de 1749, então é preciso inserir um ano posterior a ele. Tente novamente.')
        return None
    elif ano2==2025:
        print('Não é possível calcular a média suavizada de dezembro de 2025, então é preciso inserir um ano anterior a ele. Tente novamente.')
        return None
    else: 
        # obtendo a lista de médias suavizadas ------------
        listasuavizada = mediasuavizadaperiodo(ano1, ano2)[0]
        valores = [item[1] for item in listasuavizada]
        
        # identificando os índices e datas dos vales locais ------------
        vales = []
        for i in range(1, len(valores) - 1):
            if valores[i] < valores[i-1] and valores[i] < valores[i+1]:
                vales += [(i, listasuavizada[i][0])]
                # se um mês é menor que seu antecessor e que seu sucessor, é um vale.guardamos o índice 'i' porque a posição na lista funciona como uma linha do tempo em meses, simplificando as contas
        return vales
```

```python
def calculodosperiodos_picos(ano1, ano2):
    '''
    Calcula a duração dos períodos totalmente compreendidos entre os picos locais da média suavizada.
    ---------------------
    Parâmetro: ano1 (INT), ano2 (INT) anos inicial e final do período.
    ---------------------
    Retorno: lista_periodos (LIST) lista de tuplas contendo a data de início e as durações, no formato [(datadeinicio, duracaoemanos, duracaoemmeses)].
    '''
    # verificando a validade das entradas com a base de dados MENSAIS ------------
    lista_periodos = []
    if verif_doisanosMENSAIS(ano1, ano2)[0]==False:
        print(f'{" ".join(verif_doisanosMENSAIS(ano1,ano2)[1:])}')
    elif ano1==1749:
        # BUG CORRIGIDO (1): faltava o ')' para fechar o print — SyntaxError que impedia o programa inteiro de rodar.
        # BUG CORRIGIDO (4a): sem return explícito aqui, a função retornava None. O menu fazia
        # 'for p in periodos' com periodos=None -> TypeError. Adicionado return lista_periodos.
        print('Não é possível calcular a média suavizada dos 6 primeiros meses de 1749, então é preciso inserir um ano posterior a ele. Tente novamente.')
        return lista_periodos
    elif ano2==2025:
        # BUG CORRIGIDO (4a): mesmo problema de retorno None.
        print('Não é possível calcular a média suavizada de dezembro de 2025, então é preciso inserir um ano anterior a ele. Tente novamente.')
        return lista_periodos
    else:
        # obtendo os picos locais através da função envelope ------------
        picos = picos_locais_medsvz(ano1, ano2)
        if  len(picos) < 2:
            print('Não há um período inteiro entre esses dois anos')
            
        # calculando a duração entre os picos consecutivos ------------
        lista_periodos = []
        for j in range(len(picos) - 1):
            ind_inicio, (mes_inicio, ano_inicio) = picos[j]
            ind_fim, _ = picos[j+1]
            # a subtração de índices dá o total de meses; a divisão inteira por 12 nos dá a conversão para anos inteiros
            duracaoemmeses = ind_fim - ind_inicio
            duracaoemanos = duracaoemmeses // 12
            datadeinicio = f'{mes_inicio:>02}/{ano_inicio}'
            lista_periodos += [(datadeinicio, duracaoemanos, duracaoemmeses)]
        return lista_periodos
```

```python
def menu():
    # Exibe o menu de opções apenas na primeira inicialização
    exibir_opcoes()
    
    executando = True
    while executando:  # enquanto o usuário não digitar 9, o menu continua sendo executado
        erro = True
        while erro:
            try:
                # O menu aceita de 0 a 9 agora
                entrada = int(input('\nDigite a opção desejada (ou 0 para rever as opções): '))
                if entrada in range(0, 10):
                    erro = False
                else:
                    print('Opção inválida! Escolha um número de 0 a 9.')
            except ValueError:
                print('Entrada inválida! Digite apenas números inteiros.')
        
        # OPÇÃO 0: Reexibir o texto descritivo do menu
        if entrada == 0:
            exibir_opcoes()
            continue # Volta para o início do loop sem fazer mais nada

        # OPÇÃO 1: Dias sem mancha ano (a)
        elif entrada == 1:
            erro = True
            while erro:
                try:
                    ano = int(input('Insira o ano que deseja verificar (ex: 2020): '))
                    # BUG CORRIGIDO (5): verifanoin_DIARIOS(2026) retorna True mesmo que 2026
                    # tenha apenas meses 1-4. Ao chamar diassemmanchaano(2026), o loop sobre
                    # range(1,13) acessaria basededadosDIARIOS[2026][5] -> KeyError.
                    # Adicionada verificação de que TODOS os 12 meses existem.
                    if verifanoin_DIARIOS(ano)[0] == True and all(m in basededadosDIARIOS[ano] for m in range(1,13)):
                        erro = False
                    else:
                        print(f'{" ".join(verifanoin_DIARIOS(ano)[1:])} (Certifique-se de que o ano tem dados completos para todos os 12 meses.)')
                except ValueError:
                    print('Entrada inválida! Insira apenas números inteiros.')
            print('\nRESULTADO:')
            print(diassemmanchaano(ano)[1])

        # OPÇÃO 2: Dias sem manchas datas (b)
        elif entrada == 2:
            erro = True
            while erro:
                try:
                    linha = input("Insira as duas datas separadas por vírgula (ex: 12/03/2020,15/05/2021): ")
                    datas = linha.split(',')
                    data1 = datas[0].strip()
                    data2 = datas[1].strip()
                    if verif_duasdatas(data1, data2)[0] == True:
                        erro = False
                    else:
                        print(f'{" ".join(verif_duasdatas(data1, data2)[1:])}')
                except IndexError:
                    print('Entrada inválida! Certifique-se de usar a vírgula para separar as datas.')
            print('\nRESULTADO:')
            print(diassemmanchasdatas(data1, data2)[1])

        # OPÇÃO 3: Máximo de manchas datas (c)
        elif entrada == 3:
            erro = True
            while erro:
                try:
                    linha = input("Insira as duas datas separadas por vírgula (ex: 12/03/2020,15/05/2021): ")
                    datas = linha.split(',')
                    data1 = datas[0].strip()
                    data2 = datas[1].strip()
                    if verif_duasdatas(data1, data2)[0] == True:
                        erro = False
                    else:
                        print(f'{" ".join(verif_duasdatas(data1, data2)[1:])}')
                except IndexError:
                    print('Entrada inválida! Certifique-se de usar a vírgula para separar as datas.')
            print('\nRESULTADO:')
            print(maximoentredatas(data1, data2)[1])

        # OPÇÃO 4: Máximo e Mínimo Absoluto (d)
        elif entrada == 4:
            erro = True
            while erro:
                try:
                    linha = input("Insira as duas datas separadas por vírgula (ex: 12/03/2020,15/05/2021): ")
                    datas = linha.split(',')
                    data1 = datas[0].strip()
                    data2 = datas[1].strip()
                    if verif_duasdatas(data1, data2)[0] == True:
                        erro = False
                    else:
                        print(f'{" ".join(verif_duasdatas(data1, data2)[1:])}')
                except IndexError:
                    print('Entrada inválida! Certifique-se de usar a vírgula para separar as datas.')
            print('\nRESULTADO:')
            print(maxmindatas(data1, data2)[1])

        # OPÇÃO 5: Média Mensal (e)
        elif entrada == 5:
            erro = True
            while erro:
                try:
                    ano = int(input('Insira o ano que deseja verificar (ex: 1950): '))
                    # BUG CORRIGIDO (5): mesma proteção contra anos com menos de 12 meses (ex: 2026)
                    if verifanoin_DIARIOS(ano)[0] == True and all(m in basededadosDIARIOS[ano] for m in range(1,13)):
                        erro = False
                    else:
                        print(f'{" ".join(verifanoin_DIARIOS(ano)[1:])} (Certifique-se de que o ano tem dados completos para todos os 12 meses.)')
                except ValueError:
                    print('Entrada inválida! Insira apenas números inteiros.')
            print('\nRESULTADO:')
            print(mediamensal_ano(ano)[1])

        # OPÇÃO 6: Média Suavizada (g)
        elif entrada == 6:
            erro = True
            while erro:
                try:
                    linha = input("Insira os anos inicial e final separados por vírgula (ex: 1850, 1950): ")
                    anos = linha.split(',')
                    ano1 = int(anos[0].strip())
                    ano2 = int(anos[1].strip())
                    if verif_doisanosMENSAIS(ano1, ano2)[0] == True and ano1 > 1749 and ano2 < 2025:
                        erro = False
                    else:
                        print('Período inválido ou fora dos limites (use anos entre 1750 e 2024).')
                except (ValueError, IndexError):
                    print('Entrada inválida! Certifique-se de inserir dois anos inteiros separados por vírgula.')
            print('\nRESULTADO:')
            print(mediasuavizadaperiodo(ano1, ano2)[1])

        # OPÇÃO 7: Gráficos (h)
        elif entrada == 7:
            print('\nGRÁFICOS==============')
            print('1. Manchas Diárias')
            print('2. Manchas Mensais (Soma)')
            print('3. Média Mensal')
            print('4. Média Suavizada Mensal')
            
            sub_erro = True
            while sub_erro:
                try:
                    op_grafico = int(input('Escolha o gráfico desejado (1-4): '))
                    if op_grafico in [1, 2, 3, 4]:
                        sub_erro = False
                    else:
                        print('Opção inválida!')
                except ValueError:
                    print('Digite apenas inteiros.')
            
            erro_anos = True
            while erro_anos:
                try:
                    linha = input("Insira os anos inicial e final separados por vírgula (ex: 1850, 1950): ")
                    anos = linha.split(',')
                    a1 = int(anos[0].strip())
                    a2 = int(anos[1].strip())
                    # se o retorno for nulo, a função obteve êxito
                    if op_grafico == 1:
                        if plot_manchasdiarias(a1, a2)==None:
                        	erro_anos = False
                    elif op_grafico == 2:
                        if plot_manchasmensais(a1, a2)==None:
                        	erro_anos = False
                    elif op_grafico == 3:
                        if plot_mediamensal(a1, a2)==None:
                        	erro_anos = False
                    elif op_grafico == 4:
                        if plot_medsvz(a1, a2)==None:
                        	erro_anos = False
                except (ValueError, IndexError):
                    print('Entrada inválida! Insira anos inicial e final inteiros separados por vírgula.')
            import matplotlib.pyplot as plt
            plt.show()
            print('Gráfico gerado com sucesso! Feche a janela do gráfico para continuar.')

        # OPÇÃO 8: Ciclos Solares (i)
        elif entrada == 8:
            print('\n--- DURAÇÃO DOS CICLOS SOLARES ENTRE DUAS DATAS ---')
            print('1. Calcular ciclos através dos PICOS de manchas')
            print('2. Calcular ciclos através dos VALES de manchas')
            
            sub_erro = True
            while sub_erro:
                try:
                    op_ciclo = int(input('Escolha a opção (1 ou 2): '))
                    if op_ciclo in [1, 2]:
                        sub_erro = False
                    else:
                        print('Opção inválida!')
                except ValueError:
                    print('Digite apenas inteiros.')
            
            erro_anos = True
            while erro_anos:
                try:
                    linha = input("Insira os anos inicial e final de análise separados por vírgula (ex: 1750, 2024): ")
                    anos = linha.split(',')
                    a1 = int(anos[0].strip())
                    a2 = int(anos[1].strip())
                    if verif_doisanosMENSAIS(a1, a2)[0] == True:
                        erro_anos = False
                    else:
                        print('Período inválido.')
                except (ValueError, IndexError):
                    print('Entrada inválida!')
            
            if op_ciclo == 1:
                periodos = calculodosperiodos_picos(a1, a2)
            else:
                periodos = calculodosperiodos_vales(a1, a2)
                
            print('\nRESULTADO - DURAÇÃO DOS CICLOS')
            print('A lista abaixo mostra os ciclos no formato: (Data de Início, Duração em Anos Inteiros, Duração Total em Meses)')
            for p in periodos:
                print(f"Início: {p[0]} | Duração: {p[1]} anos (~{p[2]} meses)")

        # OPÇÃO 9: Sair de vez do programa
        elif entrada == 9:
            print('Encerrando o programa. Até logo!')
            executando = False
```

## 6. CONCLUSÕES

Os resultados obtidos aqui diferem dos oficiais, principalmente em função do modelo estatístico adotado. Como os dados são muito ruidosos, é necessário realizar um tratamento excelente para obter resultados com alta precisão; aqui, a qualidade do tratamento é limitada pelo desconhecimento de modelos estatísticos mais robustos e de métodos computacionais mais adequados e sofisticados. Sendo assim, é esperado que alguns resultados converjam e outros diverjam dos oficiais (dentro de uma margem aceitável).
Essa discrepância de complexidade é nítida no uso da média suavizada como o dado menos ruidoso; certamente, há tratamentos mais precisos que ela, envolvendo estatística mais sofisticada, que certamente foram usados pelas instituições de referência.

As flutuações estatísticas nos dados analisados resultam da soma de vários fatores:
Comportamento diferencial dos processos termodinâmicos solares: descrever um sistema termodinâmico qualquer já é, por natureza, razoavelmente desafiador, especialmente quando tratamos de interiores estelares. As temperaturas relativamente altas, a quantidade exorbitante de massa e energia envolvidas nos processos do interior estelar fazem com que o comportamento que observamos não seja sempre regular. Por quaisquer motivos, como uma interação gravitacional mais intensa com um cometa num dado período, ou uma ejeção de massa coronal anormal (causada por fatores desconhecidos) podem afetar ligeiramente o balanço de energia e pressão no sol, resultando no surgimento de mais ou menos manchas do que o previsto num certo período. Em suma, prever a hidrodinâmica do sol não é trivial, então naturalmente nossas expectativas irão diferir dos dados observados eventualmente.
Nebulosidade de cada dia: as observações podem ser impedidas ou afetadas pelas condições afmosféricas do dia. Nuvens muito densas inviabilizam observações cientificamente razoáveis, e nuvens não tão densas podem enviesar as medidas, já que a nitidez da obserbação diminui. Assim, a ausência de dados para certos períodos, especialmente comum nos anos do século XIX, torna as informações desses anos menos acuradas, além de que dados enviesados em função da nebulosidade também afetam as conclusões.
Subjetividade das contagens: além da influência apreciável do fator supracitado, as contagens não são absolutas, mas sim subjetivas. As contagens não são exatamente quantas manchas foram observadas naquele dia, mas sim o número de Wolf para aquele dia.
O cálculo do número de Wolf depende, entre outras informações, de quantos grupos de manchas foram observados. É difícil definir formalmente e indistinguivelmente o que é considerado um grupo de manchas, então os dados são relativos ao julgamento dos indivíduos responsáveis pelas medidas. Para um observador, um grupo poderia ser subdividido em dois, e para outro, não. Isso é especialmente agravado quando a quantidade de observadores é pequena e quando a qualidade da observação é afetada, seja por condições atmosféricas ou por limitação na resolução do equipamento utilizado. Portanto, é importante levar em conta que as medidas não são absolutas.
