```python
from random import random
import sympy as sp
import numpy as np 
from scipy.integrate import trapezoid, simpson
import time

def montecarlo(dominio,funcao, a, b):
    '''
    Função que calcula a área sob a curva de uma dada função matemática passada como argumento, dados seu domínio e os limites de integração a e b (a<b). 
    Na formulação utilizada, a razão counter_in/counter_total nos dá a probabilidade média de que, para cada x, um y aleatório seja menor que f(x), ou seja, não é a área sob a curva. Para encontrá-la,
    basta multiplicar essa razão pela área do retângulo em que estamos aplicando Monte Carlo.
    '''
    int_rng = np.array([x for x in dominio if x>a and x<b]) # definindo o intervalo de integração de acordo com o domínio da função
    counter_in = 0
    counter_total = 0
    y_max = max(funcao(int_rng))
    for x in int_rng:
        y_aleat = y_max*random()
        counter_total += 1
        if (y_aleat <= funcao(x)):
            counter_in += 1
    area = (counter_in/counter_total)*(b-a)*y_max #corrigindo com a área do retângulo utilizado para ter a devida precisão
    print(f"Monte Carlo - área = {area:.5f}")
    return area

def trapz(x,y):
    area = trapezoid(y,x)
    print (f"Trapézio - área = {area:.5f}")
    return area
def simps(x,y):
    area = simpson(y,x)
    print (f"Simpson - área = {area:.5f}")
    return area
def erro_mc(metodo, area1,area2): #  Comparando o resultado de Monte Carlo com o método 'metodo'
    print(f"Erro (Monte Carlo - {metodo}): Absoluto = {area1-area2:.4f}, relativo = {(area1-area2)/area2:+.4e}")
def erro_ex(metodo, area1,area2): #  Comparando o resultado do método 'metodo' com o valor exato obtido via Simpy
    print(f"Erro ({metodo} - exato): Absoluto = {area1-area2:.4f}, relativo = {(area1-area2)/area2:+.4e}")

```

```python
# --- Definição das 4 Funções no Intervalo [0, 1] ---

dominio = np.linspace(0,1,500)

funcoes = [
    {"nome": "1. f(x) = 3/cosh(x)", 
     "np": lambda x: 3/np.cosh(x), 
     "sp": lambda x: 3/sp.cosh(x)},
    
    {"nome": "2. f(x) = x*tan(x)", 
     "np": lambda x: x*(np.tan(x)), 
     "sp": lambda x: x*(sp.tan(x))},
    
    {"nome": "3. f(x) = x/(x^4+1)", 
     "np": lambda x: x/(1+x**4), 
     "sp": lambda x: x/(1+x**4)},
    
    {"nome": "4. f(x) = exp(x**3) / (1 + x^2)", 
     "np": lambda x: 1 / (1 + x**2), 
     "sp": lambda x: 1 / (1 + x**2)}
]

# --- Executando os testes ---

x_sym = sp.Symbol('x')
dominio = np.linspace(0, 1, 200) # domínio das 4 funções
print("AVALIAÇÃO DOS ERROS\n")
for f in funcoes:
    print("=" * 60)
    print(f"{f['nome']}")
    
    # 1. Valor Exato via SymPy
    exato = float(sp.integrate(f["sp"](x_sym), (x_sym, 0, 1)).evalf())
    print(f"Valor Exato (SymPy): {exato:.5f}\n")
    
    # 2. Avaliação da função nos pontos discretos
    y = f["np"](dominio)
    
    # 3. Cálculo das áreas
    a_trap = trapz(dominio, y)
    a_simp = simps(dominio, y)
    a_mc   = montecarlo(dominio, f["np"], a=0, b=1)
    
    # 4. Comparação dos erros com o exato
    erro_ex("Trapézio", a_trap, exato)
    erro_ex( "Simpson", a_simp, exato)
    erro_ex("Monte Carlo", a_mc, exato)
    # Comparação com Monte Carlo
    erro_mc("Trapézio", a_trap, a_mc)
    erro_mc("Simpson", a_simp, a_mc)

print("\n\nDIFERENÇA NOS TEMPOS DE EXECUÇÃO \n")
for f in funcoes:
    y = f["np"](dominio)
    # Trapézio
    stt = time.time()  # start time trapezoid
    a_trap = trapz(dominio, y)
    tempo_t = time.time()-stt

    # Simpson
    sts = time.time() # start time simpson
    a_simp = simps(dominio, y)
    tempo_s = time.time()-sts
    # Monte Carlo
    stmc = time.time() # start time mc
    a_mc   = montecarlo(dominio, f["np"], a=0, b=1)
    tempo_mc = time.time()-stmc
    print(f'Para a função {f["nome"]}:')
    print(f'Tempo (Trapézio): {tempo_t:.9f}')
    print(f'Tempo (Simpson): {tempo_s:.9f}')
    print(f'Tempo (Monte Carlo): {tempo_mc:.9f}')
    print('='*60)

```
