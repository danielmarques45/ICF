```python
import numpy as np
import time
import scipy.interpolate as itp
import scipy.optimize as opt

# Domínios
xa = np.linspace(-1, 1, 200)
xb = np.linspace(-4, 4, 300)
xc = np.linspace(-4, 4, 300)
xd = np.linspace(-4, 4, 300)
xe = np.linspace(0.01, 5, 300)
xf = np.linspace(0.01, 5, 300)

letras = [
# a
{'x': xa,
'y': 2*xa - np.cos(xa)},

#b
{'x': xb,
'y': xb**3 - 4*xb**2 - xb + 6},

#c
{'x':xc,
'y': xc*5 + 3*xc**4 - 9*xc**3 + 20*xc -12},

#d
{'x':xd,
'y':np.exp(2*xd-1) + xd},

#e
{'x': xe,
'y': np.log(xe) + xe - 2},

#f
{'x':xf,
'y':np.tanh(xf-1) - np.log(xf)}
]

def mud_sinal(dados):
    '''
    Função que, recebendo um conjunto de dados como argumento, checa se ha inversão de sinal em algum ponto.
    Se não, retorna False; se sim, retorna True; e retorna o índice da troca de sinal (se a troca acontece nos índices 120 e 121 de y, basta utilizar esses índices em x para prosseguir)
    '''
    mudanca = False
    i = 0
    loop = True
    while loop:
        if i==(len(dados)-1):
            loop=False
        elif (dados[i]*dados[i+1])<0:
            print('Índice do negativo:', i)
            mudanca = True
            loop = False
        else:
            i+=1
    return mudanca,i

def raizes(x,y): # função única que faz tudo 
    indice = mud_sinal(y)[1]
    func = itp.interp1d(x,y,kind='cubic')
    stbi = time.time()
    r1 = opt.bisect(lambda xi: func(xi), x[indice], x[indice+1])
    tempo_bi = time.time()-stbi
    print('Raiz da função (bissecção): ', r1)
    print('Tempo de execução (bissecção): ', tempo_bi)
    stbr = time.time()
    r2 = opt.brentq(lambda xi: func(xi), x[indice],x[indice+1])
    tempo_br = time.time()-stbr
    print('Raiz da função (Brent): ', r2)
    print('Tempo de execução (Brent): ', tempo_br)

let = ['A','B','C','D','E','F']
for i in range(len(letras)):
        print('='*45)
        print('LETRA ', let[i])
        raizes(letras[i]['x'],letras[i]['y'])

# Método de Brent foi mais rápido todas as vezes; a precisão dos dois métodos diverge na 12ª casa decimal
```
