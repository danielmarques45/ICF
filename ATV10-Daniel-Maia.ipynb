```python
# Bubble Sort
def bubblesort(lst):
    for iter_num in range(len(lst) - 1, 0, -1):
        trocou = False
        for idx in range(iter_num):
             if lst[idx] > lst[idx + 1]:
                lst[idx], lst[idx + 1] = lst[idx + 1], lst[idx]
                trocou = True
        if not trocou: # Interrompe o processo caso percorra a lista toda e não efetue nenhuma troca (= já está ordenada) 
            break
    return lst



# Merge Sort

def merge(left, right):
    res = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:      # <= mantém a estabilidade
            res.append(left[i]); i += 1
        else:
            res.append(right[j]); j += 1
    res.extend(left[i:])
    res.extend(right[j:])
    return res

def merge_sort(unsorted_list):
    if len(unsorted_list) <= 1:
        return unsorted_list
# Find the middle point and devide it
    middle = len(unsorted_list) // 2
    left_list = unsorted_list[:middle]
    right_list = unsorted_list[middle:]

    left_list = merge_sort(left_list)
    right_list = merge_sort(right_list)
    lista = merge(left_list, right_list)
    return lista 




# Insertion Sort
def insertion_sort(InputList):
    for i in range(1, len(InputList)):
        j = i-1
        nxt_element = InputList[i]
# Compare the current element with next one

        while (j >= 0) and (InputList[j] > nxt_element):
            InputList[j+1] = InputList[j]
            j=j-1
        InputList[j+1] = nxt_element
    return InputList


# Shell Sort
def shellSort(input_list):
    
    gap = len(input_list) // 2
    while gap > 0:

        for i in range(gap, len(input_list)):
            temp = input_list[i]
            j = i
# Sort the sub list for this gap

            while j >= gap and input_list[j - gap] > temp:
                input_list[j] = input_list[j - gap]
                j = j-gap
            input_list[j] = temp

# Reduce the gap for the next element

        gap = gap//2
    return input_list


# Selection Sort
def selection_sort(input_list):

    for idx in range(len(input_list)):

        min_idx = idx
        for j in range( idx +1, len(input_list)):
            if input_list[min_idx] > input_list[j]:
                min_idx = j
# Swap the minimum value with the compared value

        input_list[idx], input_list[min_idx] = input_list[min_idx], input_list[idx]
    return input_list

```

```python
import numpy as np
import random
import time

# Gerando o conjunto totalmente desorganizado
n = 100000
desorg = [random.randint(0,n) for _ in range(n)]

# Gerando o conjunto parcialmente desorganizado
org = [a for a in range(1,n+1)]
trocas = 5000
for _ in range(trocas):
    i = random.randint(0, n - 2)
    j = min(n - 1, i + random.randint(1, 10))
    org[i], org[j] = org[j], org[i]



# Ordenando e contando o tempo para organizar as duas listas

ti_bubble = time.time()
ld_bubble = bubblesort(desorg) # lista desorganizada bubble
lo_bubble = bubblesort(org) # lista organizada bubble
tempo_bubble = time.time()  - ti_bubble
print(f'Tempo bubble: {tempo_bubble:.6f} segundos')


ti_merge = time.time()
ld_merge = merge_sort(desorg)
lo_merge = merge_sort(org)
tempo_merge = time.time()  - ti_merge
print(f'Tempo Merge: {tempo_merge:.6f} segundos')

ti_insertion = time.time()
ld_insertion = insertion_sort(desorg)
lo_insertion = insertion_sort(org)
tempo_insertion = time.time()  - ti_insertion
print(f'Tempo Insertion: {tempo_insertion:.6f} segundos')


ti_shell = time.time()
ld_shell = shellSort(desorg)
lo_shell = shellSort(org)
tempo_shell = time.time()  - ti_shell
print(f'Tempo Shell: {tempo_shell:.6f} segundos')

ti_selection = time.time()
ld_selection = selection_sort(desorg)
lo_selection = selection_sort(org)
tempo_selection = time.time()  - ti_selection
print(f'Tempo Selection: {tempo_selection:.6f} segundos')




```
