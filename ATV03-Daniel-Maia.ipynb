```python
# Atividade 1 - Ajuste de uma curva do tipo : y = a * sin(b*x + c)
# Usando curve_fit from Scipy, modifique os exemplos dados na aula anterior 
# para fazer o ajuste pedido


from scipy import interpolate as itp
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5,5,15)
y = 4*np.sin(x+15)-np.abs(x*np.cos(3*x))
ruido = 0.1*np.random.normal(size=y.size)
y += ruido
f = itp.interp1d(x, y)
xnew = np.linspace(-5, 5, 45)
ynew = f(xnew)
plt.plot(x, y, 'o', xnew, ynew, 'x')
plt.show()


```

```python
# Atividade 2 - Ajuste de uma curva do tipo : y = a/( (x - b)**2 + c )
# Usando curve_fit from Scipy, modifique os exemplos dados na aula anterior 
# para fazer o ajuste pedido

x = np.linspace(1,10,15)
y = np.exp(x/np.pi)*np.log10(np.abs(x/np.cos(x)))
ruido = 0.1*np.random.normal(size=y.size)
y += ruido
f1 = itp.interp1d(x,y,kind='linear')
f2 = itp.interp1d(x,y,kind='quadratic')
f3 = itp.interp1d(x,y,kind='cubic')
xnew = np.linspace(1,10,60)
plt.plot(x,y,'s', markersize='7')
plt.plot(xnew,f1(xnew),'-.',lw=1.5)
plt.plot( xnew, f2(xnew),'.',lw=1.5)
plt.plot(xnew, f3(xnew),'--',lw=1.5)
plt.legend(['data', 'linear', 'quadratic','cubic'])
plt.show()
```

```python
# Atividade 3 - Ajuste de uma curva do tipo : y = a* sin(b*x)*sqrt(x+c)
# Usando curve_fit from Scipy, modifique os exemplos dados na aula anterior 
# para fazer o ajuste pedido

x = np.linspace(1,10,15)
y = 2/( (x - 3)**2 + 1 ) - np.tanh(3*x**np.e)
ruido = 0.06*np.random.normal(size=y.size)
y += ruido
f1 = itp.interp1d(x,y,kind='linear')
f2 = itp.interp1d(x,y,kind='quadratic')
f3 = itp.interp1d(x,y,kind='cubic')
xnew = np.linspace(1,10,60)
plt.plot(x,y,'s', markersize='7')
plt.plot(xnew,f1(xnew),'-.',lw=1.5)
plt.plot( xnew, f2(xnew),'.',lw=1.5)
plt.plot(xnew, f3(xnew),'--',lw=1.5)
plt.legend(['data', 'linear', 'quadratic','cubic'])
plt.show()
```
