```python
# Definindo as funções

# Indirect Method - Jacobi
#
import numpy as np
from numpy.random import randint, seed
from numpy import array, zeros, diag, diagflat, dot
from time import time

def jacobi(A,b,convergence=1e-5,N=100000,xi=None):

    """Solves the equation Ax=b via the Jacobi iterative method."""
    # Create an initial guess if needed                                                                                                                                                            
    if xi is None:
        xi = zeros(len(A[0]))

    # Create a vector of the diagonal elements of A                                                                                                                                                
    # and subtract them from A                                                                                                                                                                     
    D = diag(A)
    R = A - diagflat(D)
    xa = (b - dot(R,xi)) / D
    conv = iter = False
    c = 1
    while not(conv or iter):
          x = (b - dot(R,xa)) / D
          if max(abs(x-xa))<convergence and c<N-1:
            result = x
            conv = True
          elif c<N-1:
            c += 1
            xa = x # Redefining xa and restarting the loop
          else:
            print('No convergence achieved, try again.') # The loop has ended
            result = None
            iter = True
    return result

def gaussian_elimination(A, b):
  # Cria a matriz aumentada [A | b]
  n = len(b)
  M = np.hstack([A, b.reshape(-1, 1)]).astype(float)

  # Eliminação progressiva (Forward Elimination)
  for i in range(n):
    # Pivoteamento parcial simples para estabilidade
    max_row = i + np.argmax(np.abs(M[i:, i]))
    if i != max_row:
      M[[i, max_row]] = M[[max_row, i]]

    if M[i, i] == 0:
      raise ValueError('A matriz é singular ou não possui solução única.')

    # Zera os elementos abaixo do pivô
    for j in range(i + 1, n):
      factor = M[j, i] / M[i, i]
      M[j, i:] -= factor * M[i, i:]

  # Substituição regressiva (Back Substitution)
  x = np.zeros(n)
  for i in range(n - 1, -1, -1):
    x[i] = (M[i, -1] - np.dot(M[i, i + 1 : n], x[i + 1 : n])) / M[i, i]

  return x


# Criando as matrizes
def matrizquasediagonal(semente):
    seed(semente)
    A = zeros((50,50), dtype=int)
    diagonal = 100*randint(1,20,size=50)
    np.fill_diagonal(A,diagonal)
    for _ in range(100): # adicionaremos 250 elementos não nulos fora da diagonal
        i,j = randint(0,50,size=2)
        if i!=j and A[i,j]==0:
            A[i,j]=randint(0,50) # como a diagonal deve ser dominante, diminuímos a probabilidade de o elemento da diagonal não ser dominante na sua linha
    return A
seed(30)
inimat = time()
Arand = randint(0,501, size=(50,50)) # limite superior do randint é exclusivo
Brand = randint(0,501, size=50)
Aqdg = matrizquasediagonal(semente=32)
Bqdg = randint(0,501, size=50)
chutejacobi = randint(0,501, size=50)
tempomat = time()-inimat
print('Tempo para gerar as matrizes: ', tempomat)


# Medindo o tempo gasto para resolução

strg = time()
randg = gaussian_elimination(Arand, Brand)
temporgauss = time() - strg
stqg =time()
qdgg = gaussian_elimination(Aqdg,Brand)
tempoqgauss = time() - stqg

strj = time()
randj = jacobi(Arand,Brand,xi=chutejacobi)
temporjacobi = time() - strj
stqj = time()
qdgj = jacobi(Aqdg,Brand, xi=chutejacobi)
tempoqjacobi = time()-stqj

print(f'\nTempo para a resolução - Matriz aleatória: Gauss = {temporgauss:.7f} // Jacobi: {temporjacobi:.7f}')
print(f'Tempo para a resolução - Matriz quase diagonal: Gauss = {tempoqgauss:.7f} // Jacobi: {tempoqjacobi:.7f}')
print('\nSOLUÇÕES' + '='*80)
print(f'Aleatória \n Gauss: {randg}\n Jacobi: {randj}')
print(f'\nQuase diagonal \n Gauss: {qdgg}\nJacobi: {qdgj}\n São iguais? {'Sim' if np.allclose(qdgj,qdgg)==True else 'Não'}')
print('Perceba que, apesar da velocidade na execução, o método de Jacobi não alcançou convergência após cem mil iterações na matriz totalmente aleatória, como esperado.')
```
